### Support Vector Machines (SVM) and Kernels

#### Purpose:
- SVMs create a **hyperplane** to separate data points, making them useful for **classification tasks**.

---

#### Hard-Margin SVM
- Works well if the data is **linearly separable**
- Optimization problem:
  $$\min_{w,b} \frac{1}{2} \|w\|^2 \quad \text{subject to} \quad y_i(w^T x_i + b) \geq 1$$

**Notes**:
- Minimizing ||w||^2 is equivalent to minimizing ||w||; the 1/2 is added for **mathematical convenience**, making derivatives cleaner.
- The goal is to **maximize the geometric margin** between classes.

---

#### Representer Theorem
- Tells us that the optimal solution w can always be expressed as a linear combination of the training points:

$$w = \sum_{i=1}^{m} \alpha_i y_i x_i$$

- No need for extra points outside the training set.
- This is key for moving from **primal to dual** formulation.


----


#### Dual Optimization Problem
- Subsitute $w = \sum_{i=1}^{m} \alpha_i y_i x_i$ into the primal (hard-margin SVM formula) and simplify to get:
    
  $$\max_{\alpha} \sum_{i=1}^{m} \alpha_i - \frac{1}{2} \sum_{i=1}^{m} \sum_{j=1}^{m} \alpha_i \alpha_j y_i y_j \langle x_i, x_j \rangle$$

Subject to:
    $$\alpha_i \geq 0, \quad \sum_{i=1}^m \alpha_i y_i = 0$$

- Only **support vectors** ($\alpha_i \geq 0$) influence the solution.
- Only $\alpha_i$ and $b$ are found, prediction for a new point:
  $$f(x) = \text{sign}\left(\sum_i \alpha_i y_i \langle x_i, x \rangle + b\right)$$

----

#### Kernel Trick
- Replace the dot product $\langle x_i, x_j \rangle$ with a kernel function:

    $$K(x,z)=\phi(x)^T\phi(z)$$

- Example: polynomial kernel of degree 2: $K(x,z)=(x^Tz)^2$
- Common kernels:
    - **Linear**: $K(x,z)=x^Tz$.
    - **Polynomial**: $K(x,z) = (x^Tz + c)^d$
    - **RBF/Gaussian**: $K(x,z)=exp(-\|x-z\|^2/2\sigma^2)$

- Rule of thumb: start with standard kernels; custom kernels are only needed for **specialized domains** (strings, graphs, images).
- With kernels, the prediction becomes:

    $$f(x)=\text{sign}\left(\sum_i \alpha_i y_i K(x_i, x)+b\right)$$

----

#### Soft-Margin SVM
- Allows **misclassifications / margin violations** for noisy or non-linearly separable data.
- Introduces **slack variables $\xi_i$** or equivalently use **hinge loss**:

$$min_{w,b} \sum_{i=1}^{m} \max(0, 1 - y_i(w^T x_i + b)) + \lambda \|w\|^2$$

- $\max(0, 1 - y_i f(x_i))$ measures **margin violations**
- $\lambda$ controls the trade-off between **maximizing margin** vs **minimizing hinge loss**
-  Smaller $\lambda$ -> tolerate more margin violations
-  Larger $\lambda$ -> enforce stricter separation, behaves like hard-margin

----

#### Key Practical Points:
- **Training**: Solve for $\alpha_i$ using **Quadratic Programming (QP)** or iterative methods like Gradient Descent
  
- **Prediction**: Only the support vectors ($\alpha_i \geq 0$) are used, making prediction efficient.

- **Soft-margin SVM** is almost always used in practice because **real-world-data is rarely perfectly separable.**

- **Kernel trick** allows SVMs to handle **high dimensional or infinite-dimensional feature spaces** without explicitly computing them.
