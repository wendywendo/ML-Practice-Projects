```python
# Import necessary modules
import pandas as pd
import numpy as np

# Visualisation
import matplotlib.pyplot as plt
import matplotlib as mpl

# For balancing data
from imblearn.over_sampling import SMOTE
```

```python
diseases_df = pd.read_csv('diseases.csv')
diseases_df.head()
```

```python
X = diseases_df.drop(['disease'], axis=1)
X.head()
```

```python
y = diseases_df['disease']
y.head()
```

```python
# Split the data
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3
)

oversample = SMOTE()
X_train, y_train = oversample.fit_resample(X_train, y_train)
```

```python
# Linear SVC, KNeighborsClassifier, SVC, Ensemble Classifiers
```

```python
C = 10
classifiers = {
    'Linear SVC': SVC(kernel='linear', C=C, probability=True, random_state=0),
    'KNN Classifier': KNeighborsClassifier(10),
    'SVC': SVC(),
    'RFST': RandomForestClassifier(n_estimators=100),
    'ADA': AdaBoostClassifier(n_estimators=100)
}
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'C': [0.01, 0.1, 1, 5, 10, 20, 100]}
grid = GridSearchCV(
    SVC(kernel='linear'),
    param_grid,
    scoring='f1_macro',
    cv=5,
    n_jobs = -1
)

grid.fit(X_train, y_train)
print(grid.best_params_)
```

```python
n_classifiers = len(classifiers)

for index, (name, classifier) in enumerate(classifiers.items()):
    classifier.fit(X_train, y_train)

    print(name)
    y_pred = classifier.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print("Accuracy for %s: %0.1f%% "% (name, accuracy * 100))
    print(classification_report(y_test, y_pred))
```

```python
# Linear SVC performs best
from sklearn.svm import SVC

linear_svc = SVC(
    kernel="linear",
    C=10,
    class_weight="balanced",
    random_state=42
)

linear_svc.fit(X_train, y_train)
```

```python
from sklearn.metrics import f1_score

y_pred = linear_svc.predict(X_test)

# print(classification_report(y_test, y_pred))
print(f1_score(y_test, y_pred, average='weighted'))
```

```python
possible_symptoms = ['fever', 'headache', 'nausea', 'vomiting', 'fatigue', 'joint_pain', 'skin_rash', 'cough', 'weight_loss', 'yellow_eyes']

def predict_top_10_diseases(symptoms):
    input_data = [0] * len(possible_symptoms)

    for symptom in symptoms.split(','):
        symptom = symptom.strip()

        if symptom in possible_symptoms:
            idx = possible_symptoms.index(symptom)
            input_data[idx] = 1

        sample = pd.DataFrame([input_data], columns=X_train.columns)

        # Get decision scores
        scores = linear_svc.decision_function(sample)[0]

        # Get indices of top 10 scores
        top_10_idx = np.argsort(scores)[-10:][::-1]

        # Map indices to disease names and scores
        top_10_diseases = [
            (linear_svc.classes_[i], scores[i])
            for i in top_10_idx
        ]

        return top_10_diseases
```

```python
symptoms = "fever,nausea"
top_10 = predict_top_10_diseases(symptoms)

for disease, score in top_10:
    print(f"{disease:40s} confidence score: {score:.3f}")
```

```python
def top_k_accuracy(model, X_test, y_test, k=10):
    scores = model.decision_function(X_test)
    top_k_preds = np.argsort(scores, axis=1)[:, -k:]

    correct = 0
    for i, true_label in enumerate(y_test):
        true_index = np.where(model.classes_ == true_label)[0][0]
        if true_index in top_k_preds[i]:
            correct += 1

    return correct / len(y_test)
```

```python
acc = top_k_accuracy(linear_svc, X_test, y_test, k=10)
print(f"Accuracy: {acc * 100:.2f}%")
```
